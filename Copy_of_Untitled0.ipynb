**AI ChatBot for Student Support**

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

# Chatbot knowledge base (FAQs)
def get_bot_response(user_query):
    query = user_query.lower()

    if any(word in query for word in ['admission', 'apply', 'enroll', 'course']):
        print("Bot: Admissions for the new academic session are open! You can apply online through the student portal or visit the admission office between 9 AM and 4 PM.")
    elif any(word in query for word in ['fee', 'payment', 'tuition', 'due']):
        print("Bot: Fee payments can be made online via the student dashboard using net banking, UPI, or credit/debit card. Last date for semester fee submission is the 25th of this month.")
    elif any(word in query for word in ['exam', 'test', 'result', 'schedule', 'date sheet']):
        print("Bot: End-semester examination schedule has been released on the portal. Results for the previous semester are available under your student profile.")
    elif any(word in query for word in ['library', 'book', 'issue', 'fine']):
        print("Bot: The central library is open from 8 AM to 8 PM. Students can issue up to 3 books at a time for a duration of 14 days.")
    elif any(word in query for word in ['timing', 'hours', 'open', 'college time']):
        print("Bot: College administrative offices are open Monday to Saturday, 9:00 AM to 5:00 PM.")
    elif any(word in query for word in ['hello', 'hi', 'hey', 'start']):
        print("Bot: Hello! Welcome to Student Support Helpdesk. How can I help you today? (Ask about: Admission, Fees, Exams, Library)")
    elif any(word in query for word in ['bye', 'exit', 'quit']):
        print("Bot: Thank you for reaching out! Have a great day ahead.")
    else:
        print("Bot: I'm sorry, I didn't quite catch that. You can ask me about admissions, fees, exam schedules, library timings, or general college queries.")

# UI Setup using ipywidgets for Colab
print("--- Student Support AI Chatbot ---")
print("Type your question below and press Enter (Type 'bye' to exit).\n")

text_input = widgets.Text(
    value='',
    placeholder='Type your question here...',
    description='You:',
    disabled=False,
    layout=widgets.Layout(width='80%')
)

output_area = widgets.Output()

def on_submit(sender):
    query = text_input.value.strip()
    if not query:
        return

    with output_area:
        print(f"You: {query}")
        get_bot_response(query)
        print("-" * 50)

    # Clear input box for next message
    text_input.value = ''

text_input.on_submit(on_submit)

# Display widgets
display(text_input, output_area)
```
